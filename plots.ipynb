```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("empirical_results.csv")
df.head()
```

```python
plt.figure(figsize=(4, 3))
sns.lineplot(data=df, x='Files', y='Time', hue='Algorithm', markers='o')
plt.xscale('log')
plt.tight_layout()
plt.savefig('benchmark_plot.png')
plt.show()
```

```python
plt.figure(figsize=(4, 3))
sns.lineplot(data=df, x='Files', y='Time', hue='Algorithm')
plt.xlabel('Files (log scale)')
plt.ylabel('Time (log scale)')
plt.xscale('log')
plt.yscale('log')
plt.tight_layout()
plt.savefig('benchmark_plot_log.png')
plt.show()
```
